### Sampling Caveats
#### 1.Dataset Representation
The dataset may not represent all Apple products available in the global market.
Some product categories, regions, or models may have limited records.
#### 2. Time Period Limitation
The data covers products from 2020–2026, so older products and future launches are not included.
#### 3.Regional Price Differences
Launch prices and current prices may vary by country due to:
Taxes
Currency conversion
Import duties
Regional pricing strategies
#### 4.Sample Size and Distribution
Some product categories may have fewer observations compared to others.
Statistical results may be influenced by categories with more records.
#### 5.Missing or Incomplete Data
Missing values or incomplete price information can affect analysis results.
#### 6.Price Changes Over Time
Current prices may change due to:
Discounts
Sales events
Market demand
Stock availability
#### 7.Outliers
Premium products with very high prices may affect:
Mean price
Standard deviation
Distribution analysis
Data Source Reliability
Results depend on the accuracy and consistency of the original data source.
#### Conclusion: 
- "The analysis provides insights based on the available dataset, but findings should be interpreted considering limitations in sample     coverage, regional variations, and changing market conditions."
